```python
#Calculate: Total Purchase Orders Total Ordered Quantity Total Received Quantity Average Ordered Quantity Average Received Quantity
import pandas as pd

df = pd.read_csv(r"C:\Users\yadav\Documents\archive (3)\purchase_orders.csv")

summary = df.groupby(["supplier_id", "site_id"]).agg(
    Total_PO=("po_id", "count"),
    Total_Ordered=("ordered_qty", "sum"),
    Total_Received=("received_qty", "sum"),
    Avg_Ordered=("ordered_qty", "mean"),
    Max_Ordered=("ordered_qty", "max"),
    Min_Ordered=("ordered_qty", "min"),
    Avg_Received=("received_qty", "mean")
).sort_values("Total_Ordered", ascending=False)

print(summary)
```

```python
#Create a new column called Order_Status based on the following rules:
#If received_qty == ordered_qty → "Complete"
#If received_qty == 0 → "Not Received"
#Otherwise → "Partial"
import numpy as np 
import pandas as pd 
df = pd.read_csv(r"C:\Users\yadav\Documents\archive (3)\purchase_orders.csv")
df["order_status"]=np.where(df["received_qty"]==df["ordered_qty"],"complete",
                   np.where(df["received_qty"]==0,"not received","partial"))
print(df[["po_id","ordered_qty","received_qty","order_status"]].head(10))
df.groupby(["supplier_id","order_status"]).agg(Total_PO=("po_id", "count"),
    Total_Ordered=("ordered_qty", "sum"),
    Total_Received=("received_qty", "sum")).sort_values("Total_Ordered", ascending=False)
```

```python
import numpy as np
import pandas as pd
#Data Cleaning & Data Transformation

df=pd.read_csv(r"C:\Users\yadav\Documents\archive (3)\purchase_orders.csv")
df.isnull().sum()
df.isnull().sum().sum()
df[df.isnull().any(axis=1)]
df.isnull().sum()/len(df)*100.0

#Handling Missing Values

df["receipt_date"].isnull().sum()
df["receipt_date"]=df["receipt_date"].fillna('pending')
df["receipt_date"].isnull().sum()
print(df[["po_id","receipt_date"]].head(15))

#Dropping Missing Values
df=pd.read_csv(r"C:\Users\yadav\Documents\archive (3)\purchase_orders.csv")
df.shape
df = df.dropna(subset=["receipt_date"])
df.shape
df["receipt_date"].isnull().sum()
```

```python
#Removing Duplicate Records
import numpy as np
import pandas as pd
df=pd.read_csv(r"C:\Users\yadav\Documents\archive (3)\purchase_orders.csv")
df.shape
df.duplicated().sum()
df[df.duplicated()]
df=df.drop_duplicates()
df.shape
#Date Handling & Date Arithmetic
df["order_date"]=pd.to_datetime(df["order_date"])
df["promised_date"]=pd.to_datetime(df["promised_date"])
df["receipt_date"]=pd.to_datetime(df["receipt_date"])
df["Delivery_Days"] = df["receipt_date"] - df["order_date"]
df["Delay_Days"] = (df["receipt_date"] - df["promised_date"]).dt.days
print(df[["po_id","supplier_id","order_date","promised_date","receipt_date","Delivery_Days","Delay_Days"]])

#Apply Function
def classify_orders(ordered_qty):
    if ordered_qty < 100:
        return "small"
    elif 100 <= ordered_qty <300:
        return "medium"
    else :
        return "large"

df["order_size"]=df["ordered_qty"].apply(classify_orders)
print(df[["po_id","ordered_qty","order_size"]])
```
